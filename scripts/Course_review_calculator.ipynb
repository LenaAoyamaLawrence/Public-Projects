# Course review calculator
* this calculator calculates the medium, mean and standard deviation from the survey
* needed variables:
    * number of "very good"
    * number of "good"
    * number of "average"
    * number of "bad"
    * number of "very bad"

```python
# import dependencies
import pandas as pd
import numpy as np
```

```python
# initialise variables
veryGood = 11
good = 18
average = 4
bad = 0
veryBad = 0
count_response = veryGood + good + average + bad + veryBad
```

```python
# calculate score
veryGood_score = np.ones(veryGood) * 1
good_score = np.ones(good) * 2
average_score = np.ones(average) * 3
bad_score = np.ones(bad) * 4
veryBad_score = np.ones(veryBad) * 5 
```

```python
# concantenate the arrays
score_array = np.concatenate((veryGood_score, good_score, average_score, bad_score, veryBad_score))
score_array.shape
```

```python
# mean calculation
mean_score = np.mean(score_array)

std_score = np.std(score_array)

median_score = np.median(score_array)

print(f"mean score: {mean_score}, standard deviation score: {std_score}, median score: {median_score}")
```
